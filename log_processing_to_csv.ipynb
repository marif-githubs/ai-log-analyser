###convert row log file to csv


```python
import pandas as pd

file = open("./permission_issue_20260520_160313.log", "r")

content = file.read()

file.close()

data = content.splitlines()

description = []
date = []
time = []
tag = []
component = []

for text in data:
    
    parts = text.split(":", 3)

    temp = ":".join(parts[:3])

    description.append(parts[3])
    meta_parts = temp.split()
    date.append(meta_parts[0])
    time.append(meta_parts[1])
    tag.append(meta_parts[2])
    component.append(meta_parts[3])

"""print(description, end='')
print(date, end='')
print(time, end='')
print(tag, end='')
print(component, end='')"""

csv_data = {
    'date':date,
    'time':time,
    'tag':tag,
    'component':component,
    'description':description
}

df = pd.DataFrame(csv_data)
print(df.head)
df.to_csv("permission_issue_20260520_160313.log.csv", index=False)
```
